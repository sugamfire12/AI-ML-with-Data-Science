```python
import numpy as np
z = np.array([[1, 2, 3], [4, 5, 6]])

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

output = sigmoid(z)
print('input:\n', z)
print('output:\n', output)
```

```python
x = np.array([2, 3])
w = np.array([0.5, 0.8])
b = 1

z = np.dot(x, w) + b
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

output = sigmoid(z)
print("Weighted sum:", z)
print(" Sigmoid Output:", output)
```

```python
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)
linear = x
sigmoid = 1 / (1 + np.exp(-x))

plt.plot(x, linear, label='Linear')
plt.plot(x, sigmoid, label='Sigmoid')

plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.grid()
plt.show()
```

## ReLU (Rectified Linear Unit) Function

vanishing gradiet problem: is a training difficulty in training neural networks where the error signal shrinks exponentially as it propagates backward through layers, causing the weights in earlier layers to barely update or freeze complelely.

```python
x = np.array([-3, -1, 0, 1, 3])
relu = np.maximum(0, x)
print("Input:", x)
print("ReLU Output:", relu)
```

### Leaky ReLU
It allows a small negative values instead of completely making it zero.

leakyrelu(x) = x if x >= 0
                alpha if x <= 0 

```python
x = np.array([-5, -2, 0, 2, 5])
alpha = 0.01
output = np.where(x > 0, x, alpha * x)
print("Input:", x)
print("Leaky ReLU Output:", output)
```

### ELU - Exponential Linear Unit
It is another improvement over ReLU. The main difference is that ELU smoothly handles negative values instead of simply making them zero.

ELU(x) = x if x >= 0
        a(epow(x) - 1) if x <= 0
